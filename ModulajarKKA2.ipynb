```python
import pandas as pd

# Define the data_kantin dictionary (reconstructing from kernel state)
data_kantin = {
    'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
    'harga': [12000, 4000, 10000, 4000, 8000],
    'terjual': [23, 40, None, 35, 18]
}

# Create DataFrame from the dictionary
df = pd.DataFrame(data_kantin)

print("--- 5 Baris Pertama ---")
print(df.head())

print("\n--- Informasi Data (info) ---")
print(df.info())

print("\n--- Statistik Ringkas (describe) ---")
print(df.describe())

print("\n--- Ukuran Dataset (shape) ---")
print(df.shape)
```

Jawaban Tugas Analisis 3:
Analisis df.info():

Kolom yang memiliki jumlah non-null lebih sedikit dibanding total baris adalah kolom menu (4 non-null) dan kolom terjual (4 non-null) dari total 5 baris data.

```python
# Cek jumlah data kosong tiap kolom
print("Data kosong sebelum pembersihan:")
print(df.isnull().sum())

# Isi kekosongan pada kolom terjual dengan angka 0
df['terjual'] = df['terjual'].fillna(0)

# Hapus baris yang kolom menu-nya kosong
df = df.dropna(subset=['menu'])

print("\nData kosong setelah pembersihan:")
print(df.isnull().sum())
```

Jawaban Tugas Analisis 4:

Mengapa terjual diisi fillna(0)?

Karena Secara logika bisnis, jika jumlah item terjual bernilai kosong (null), asumsi yang paling masuk akal adalah item tersebut belum/tidak ada yang terjual (0 unit) pada periode tersebut. Mengisinya dengan angka 0 mempertahankan baris data tanpa merusak perhitungan agregasi numerik

Mengapa menu dihapus dropna()?

Menu adalah identitas utama/kategori dari barang yang dijual. Kita tidak bisa menebak menu apa yang terjual senilai Rp8.000 hanya dari angka saja. Membiarkan baris tanpa nama menu akan membuat analisis agregasi menu (seperti menu mana yang paling laris) menjadi tidak valid atau ambigu.

```python
# Cek jumlah baris duplikat
print("Jumlah duplikat:", df.duplicated().sum())

# Hapus baris duplikat
df = df.drop_duplicates()

# Mengubah tipe data harga menjadi integer
df['harga'] = df['harga'].astype(int)

# Tampilkan tipe data seluruh kolom
print("\nTipe data setelah penyesuaian:")
print(df.dtypes)
```

Jawaban Tugas Analisis 5:

Jumlah baris setelah drop_duplicates():
Sebelumnya terdapat 4 baris data tersisa (setelah dropna pada Latihan 4). Karena 'Es Teh' muncul 2 kali dengan data yang sama persis (harga: 4000, terjual: 40), 1 baris duplikat dihapus, sehingga jumlah baris dataset menjadi 3 baris.

```python
# 1. Filtering: Menu laris (terjual > 20)
laris = df[df['terjual'] > 20]
print("--- Menu Laris (> 20) ---")
print(laris)

# 2. Sorting: Mengurutkan berdasarkan terjual secara terbanyak (descending)
urut = df.sort_values(by='terjual', ascending=False)

# 3. Kolom Turunan: Menghitung total pendapatan
df['total_pendapatan'] = df['harga'] * df['terjual']

# 4. Agregasi / Groupby: Total pendapatan per menu
ringkasan = df.groupby('menu')['total_pendapatan'].sum()
print("\n--- Ringkasan Pendapatan per Menu ---")
print(ringkasan)
```
